```python
import pandas as pd
```

```python
df = pd.read_csv('model_behavior.csv')
df = df.drop(columns=['template_id', 'split', 'answer_prob', 'total_answer_prob', 'cot', 'valid_answer'])
df.head()
```

```python
# convert df to json of the form - qid:, question:, model_answer:, params: {a, b, c, d, e}
data = []
for index, row in df.iterrows():
    item = {}
    item['qid'] = row['question_id']
    item['question'] = row['question_text']
    item['model_answer'] = row['completion']
    item['params'] = {
        'a': row['a'],
        'b': row['b'],
        'c': row['c'],
        'd': row['d'],
        'e': row['e']
    }
    data.append(item)
```

```python
# save data to json file
import json
with open('data_sycophancy.json', 'w') as f:
    json.dump(data, f, indent=2)
```

```python
# give summary of the data
print(f'Total questions: {len(data)}')
print(f'Total Yes answers: {sum(1 for item in data if item["model_answer"].strip().lower() == "yes")}')
print(f'Total No answers: {sum(1 for item in data if item["model_answer"].strip().lower() == "no")}')
```
